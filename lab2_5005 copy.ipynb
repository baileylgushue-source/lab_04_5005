```python
import json
from pathlib import Path
import requests
```

```python
URL = "https://api.tvmaze.com/shows?page=0"
```

```python
OUTPUT = Path("summary.json")
```

```python
response = requests.get(URL, timeout=10)
```

```python
response.raise_for_status()
shows = response.json()
```

```python
show_info = ("TVMaze", len(shows))
print (show_info)
```

```python
show_names = [show["name"] for show in shows]
print(show_names)
```

```python
def fetch_shows(url):
    """Download the TV show records and return them as Python objects."""
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    return response.json()
```

```python
def shows_per_genre(records):
    """Count the number of shows in each genre."""
    genre_amount = {}

    for show in records:
        for genre in show["genres"]:
            genre_amount[genre] = genre_amount.get(genre, 0) + 1

    return genre_amount


print(f"Shows by genre: {shows_per_genre(shows)}")
```

```python
def shows_per_language(records):
    """Count the number of shows in each language."""
    language_amount = {}

    for show in records:
        language = show.get("language")

        if language is not None:
            language_amount[language] = language_amount.get(language, 0) + 1

    return language_amount


print(f"Shows by language: {shows_per_language(shows)}")
```

```python
def build_summary(records):
    """Combine the aggregations into one dict ready to write."""

    summary = {
        "source_url": URL,
        "records_processed": len(records),
        "shows_per_genre": shows_per_genre(records),
        "shows_per_language": shows_per_language(records)
    }

    return summary
print(f"Summary: {build_summary(shows)}")
```

```python
def write_summary(summary, path):
    """Write the summary to a JSON file."""
    path.write_text(json.dumps(summary, indent=2), encoding="utf-8")
```

```python
def main():
    shows = fetch_shows(URL)
    genre_amount = shows_per_genre(shows)
    language_amount = shows_per_language(shows)

print(f"Read {len(shows)} TV show records.")
print(f"Summary written to {OUTPUT}")


if __name__ == "__main__":
    main()
```
